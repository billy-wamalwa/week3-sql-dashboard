# **Step 3: Part A - Write and Explain 5 SQL Querries**

## **Query 1 - How many interns completed each domain's program?**

```python
query = """
--Identifies the number of interns who completed the programme in each domain, ranked from highest to lowest.
SELECT Domain_Applied, COUNT (*) AS completed_count
FROM interns
WHERE Completion_status ='Completed'
GROUP BY Domain_Applied
ORDER BY completed_count DESC;
"""
interns_completed = pd.read_sql_query(query, conn)

display(interns_completed)
```

## **Query 2 - What is the average hackathon score per domain?**

```python
query = """
-- Calculates the average hackathon_score for each domain, ranked from highest to lowest.
SELECT Domain_Applied, ROUND(AVG(score), 2) AS Average_Score
FROM hackathon_scores
GROUP BY Domain_Applied
ORDER BY  Average_Score DESC;
"""
average_hackathon_score = pd.read_sql_query(query, conn)

display(average_hackathon_score)
```

## **Query 3 - Which interns scored above threshold(top performers)?**

```python
query = """
-- Lists interns who scored 85 or above, ranked from highest to lowest.
SELECT i.intern_id, i.Domain_Applied, h.Score
FROM interns i
JOIN hackathon_scores h ON i.intern_id = h.intern_id
WHERE h.Score >= 85
ORDER BY h.Score DESC;
"""
top_performers = pd.read_sql_query(query, conn)

display(top_performers)
```

## **Query 4 - What Was The Conversion rate from "Applied" to "Completed" per domain?**

```python
query = """
-- Compares total applicants vs completed interns per domain to find the conversion rate
SELECT 
a.Domain_Applied,
COUNT(DISTINCT a.applicant_id) AS total_applicants,
COUNT(DISTINCT i.intern_id) AS total_completed,
ROUND(
    100.0 * COUNT(DISTINCT i.intern_id) / COUNT(DISTINCT a.applicant_id), 
    2
) AS conversion_rate_pct
FROM applicants a
LEFT JOIN interns i
ON a.Applicant_ID = i.Applicant_ID AND i.Completion_status = 'Completed'
GROUP BY a.Domain_Applied
ORDER BY conversion_rate_pct DESC;
"""
conversion_rate = pd.read_sql_query(query, conn)

display(conversion_rate)
```

## **Query 5 - What were the top 3 Universities with the highest scoring interns?**

```python
query = """
-- Identifies the top 3 universities with the highest average hackathon score among completed interns. 
SELECT
    a.University_Name,
    COUNT(DISTINCT i.intern_id) AS total_interns,
    ROUND(AVG(h.Score), 2) AS average_score
FROM applicants a
INNER JOIN interns i
    ON a.Applicant_ID = i.Applicant_ID
INNER JOIN hackathon_scores h
    ON i.intern_id = h.intern_id
WHERE i.Completion_status = 'Completed'
GROUP BY a.University_Name
ORDER BY average_score DESC
LIMIT 3;
"""

top_universities = pd.read_sql_query(query, conn)

display(top_universities)
```
